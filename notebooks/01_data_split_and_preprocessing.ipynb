```python
import pandas as pd
import joblib
import category_encoders as ce
from sklearn.preprocessing import StandardScaler

import sys
sys.path.append("../")
from src import *
```

```python
train_data = pd.read_csv("../data/training.csv")
X = train_data.drop(columns='IsBadBuy')
y = train_data['IsBadBuy']
```

```python
X_train, X_valid, X_test, y_train, y_valid, y_test = date_split(X, y, date_col="PurchDate", train_size=1/3, val_size=1/3)
```

```python
cat_cols = X_train.select_dtypes(include="object").columns.to_list()
num_cols = X_train.select_dtypes(exclude="object").columns.to_list()
```

```python
# Категориальные
X_train[cat_cols] = X_train[cat_cols].fillna("Missing")
X_valid[cat_cols] = X_valid[cat_cols].fillna("Missing")
X_test[cat_cols]  = X_test[cat_cols].fillna("Missing")

# Числовые
for col in num_cols:
    median = X_train[col].median()
    X_train[col] = X_train[col].fillna(median)
    X_valid[col] = X_valid[col].fillna(median)
    X_test[col]  = X_test[col].fillna(median)
```

```python
encoder = ce.CountEncoder(cols=cat_cols)

X_train_enc = encoder.fit_transform(X_train).drop(['PurchDate'], axis=1)
X_valid_enc = encoder.transform(X_valid).drop(['PurchDate'], axis=1)
X_test_enc  = encoder.transform(X_test).drop(['PurchDate'], axis=1)
```

```python
joblib.dump(X_train_enc, '../data/01/X_train_raw.pkl')
joblib.dump(X_valid_enc, '../data/01/X_valid_raw.pkl')
joblib.dump(X_test_enc, '../data/01/X_test_raw.pkl')
```

```python
scaler = StandardScaler()

X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train_enc),
    columns=X_train_enc.columns,
    index=X_train_enc.index
)

X_valid_scaled = pd.DataFrame(
    scaler.transform(X_valid_enc),
    columns=X_valid_enc.columns,
    index=X_valid_enc.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test_enc),
    columns=X_test_enc.columns,
    index=X_test_enc.index
)
```

```python
joblib.dump(X_train_scaled, "../data/01/X_train.pkl")
joblib.dump(y_train, "../data/01/y_train.pkl")

joblib.dump(X_valid_scaled, "../data/01/X_valid.pkl")
joblib.dump(y_valid, "../data/01/y_valid.pkl")

joblib.dump(X_test_scaled, "../data/01/X_test.pkl")
joblib.dump(y_test, "../data/01/y_test.pkl")

joblib.dump(encoder, "../data/01/count_encoder.pkl")
joblib.dump(scaler, "../data/01/scaler.pkl")
```
