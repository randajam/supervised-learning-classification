```python
import joblib
import pandas as pd

from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

import sys
sys.path.append("../")
from src import *
```

```python
X_train = joblib.load('../data/01/X_train.pkl')
y_train = joblib.load('../data/01/y_train.pkl')

X_valid = joblib.load('../data/01/X_valid.pkl')
y_valid = joblib.load('../data/01/y_valid.pkl')
```

```python
def gini_score(y_true, y_pred_proba):
    return 2 * roc_auc_score(y_true, y_pred_proba) - 1
```

```python
lr = LogisticRegression(max_iter=1000, n_jobs=-1, random_state=21)

lr.fit(X_train, y_train)

lr_pred_valid = lr.predict_proba(X_valid)[:, 1]
joblib.dump(lr_pred_valid, '../data/02/lr_pred_valid.pkl')

lr_gini = gini_score(y_valid, lr_pred_valid)
print("Logistic Regression Gini:", lr_gini)
```

```python
gnb = GaussianNB()
gnb.fit(X_train, y_train)

gnb_pred_valid = gnb.predict_proba(X_valid)[:, 1]
joblib.dump(gnb_pred_valid, '../data/02/gnb_pred_valid.pkl')

gnb_gini = gini_score(y_valid, gnb_pred_valid)
print("GaussianNB Gini:", gnb_gini)
```

```python
knn = KNeighborsClassifier(n_neighbors=15, weights="distance")
knn.fit(X_train, y_train)

knn_pred_valid = knn.predict_proba(X_valid)[:, 1]
joblib.dump(knn_pred_valid, '../data/02/knn_pred_valid.pkl')

knn_gini = gini_score(y_valid, knn_pred_valid)
print("KNN Gini:", knn_gini)
```

```python
results = pd.DataFrame({
    "Model": ["Logistic Regression", "GaussianNB", "KNN"],
    "Gini": [lr_gini, gnb_gini, knn_gini]
}).sort_values("Gini", ascending=False)

results
```
