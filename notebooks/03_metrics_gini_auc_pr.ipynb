```python
import joblib
import pandas as pd
from sklearn.metrics import roc_auc_score

import sys
sys.path.append("../")
from src import *
```

```python
X_valid = joblib.load("../data/01/X_valid.pkl")
y_valid = joblib.load("../data/01/y_valid.pkl")

# Предсказания моделей (probabilities)
lr_pred_valid = joblib.load("../data/02/lr_pred_valid.pkl")
gnb_pred_valid = joblib.load("../data/02/gnb_pred_valid.pkl")
knn_pred_valid = joblib.load("../data/02/knn_pred_valid.pkl")
```

```python
def gini_score(y_true, y_pred):
    return 2 * roc_auc_score(y_true, y_pred) - 1

print(f"My roc_auc_score: {roc_auc_manual(y_valid, lr_pred_valid)}")
print(f"Sklearn roc_auc_score: {roc_auc_score(y_valid, lr_pred_valid)}")

print(f"\nMy gini: {gini_manual(y_valid, lr_pred_valid)}")
print(f"Sklearn gini: {gini_score(y_valid, lr_pred_valid)}")
```

```python
models = {
    "Logistic Regression": lr_pred_valid,
    "GaussianNB": gnb_pred_valid,
    "KNN": knn_pred_valid
}

results = []

for name, pred in models.items():
    gini = gini_manual(y_valid, pred)
    roc_auc = roc_auc_manual(y_valid, pred)
    pr_auc = pr_auc_manual(y_valid, pred)
    results.append([name, gini, roc_auc, pr_auc])

import pandas as pd

results_df = pd.DataFrame(results, columns=["Model", "Gini", "ROC AUC", "PR AUC"])
results_df = results_df.sort_values("Gini", ascending=False)
results_df
```
