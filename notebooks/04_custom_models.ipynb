```python
import joblib
import pandas as pd
from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

import sys
sys.path.append('../')
from src import *
```

```python
X_train = joblib.load('../data/01/X_train.pkl')
y_train = joblib.load('../data/01/y_train.pkl')

X_valid = joblib.load('../data/01/X_valid.pkl')
y_valid = joblib.load('../data/01/y_valid.pkl')
```

```python
# Logistic Regression
sk_lr = LogisticRegression()
sk_lr.fit(X_train, y_train)
lr_pred = sk_lr.predict_proba(X_valid)[:, 1]
print("Sklearn Logistic Regression Gini:", gini_score(y_valid, lr_pred))

cl_lr = CustomLogisticRegression(lr=0.01, n_iter=1000)
cl_lr.fit(X_train.values, y_train.values)
lr_pred = cl_lr.predict_proba(X_valid)
print("Custom Logistic Regression Gini:", gini_score(y_valid, lr_pred))
```

```python
# KNN (можно использовать подвыборку для скорости)
sk_knn = KNeighborsClassifier(n_neighbors=15)
sk_knn.fit(X_train[:2000], y_train[:2000])
knn_pred = sk_knn.predict_proba(X_valid)[:, 1]
print("Sklearn KNN Gini:", gini_score(y_valid, knn_pred))

cl_knn = CustomKNN(n_neighbors=15)
cl_knn.fit(X_train.values[:2000], y_train.values[:2000])  # подвыборка для скорости
knn_pred = cl_knn.predict_proba(X_valid.values)      # подвыборка validation
print("Custom KNN Gini:", gini_score(y_valid.values, knn_pred))
```

```python
# GaussianNB
sk_gnb = GaussianNB()
sk_gnb.fit(X_train, y_train)
gnb_pred = sk_gnb.predict_proba(X_valid)[:, 1]
print("Sklearn GaussianNB Gini:", gini_score(y_valid, gnb_pred))

cl_gnb = CustomGaussianNB()
cl_gnb.fit(X_train.values, y_train.values)
gnb_pred = cl_gnb.predict_proba(X_valid.values)[:, 1]
print("Custom GaussianNB Gini:", gini_score(y_valid, gnb_pred))
```

```python
result = pd.DataFrame({
    'Algorithm': ['Logistic Regression', 'KNN', 'GaussianNB'],
    'gini(not FE)' : [gini_score(y_valid, lr_pred),gini_score(y_valid, knn_pred),gini_score(y_valid, gnb_pred)]
})
```
